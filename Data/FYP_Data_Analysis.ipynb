```python
from scipy.stats import ttest_ind

male_height = df[df["Sex"] == "Male"]["Height_cm"]
female_height = df[df["Sex"] == "Female"]["Height_cm"]

t_stat, p_value = ttest_ind(male_height, female_height)

print("t-statistic:", t_stat)
print("p-value:", p_value)
```

```python
from scipy.stats import levene

levene_stat, levene_p = levene(male_height, female_height)

print("Levene statistic:", levene_stat)
print("Levene p-value:", levene_p)
```

```python
import numpy as np

n_male = len(male_height)
n_female = len(female_height)

pooled_sd = np.sqrt(
    ((n_male - 1) * male_height.var(ddof=1) +
     (n_female - 1) * female_height.var(ddof=1))
    / (n_male + n_female - 2)
)

cohens_d = (male_height.mean() - female_height.mean()) / pooled_sd

print("Cohen's d:", cohens_d)
```

```python
# Mean height was 160.77 ± 6.27 cm among females and 173.69 ± 6.75 cm among males. 
# An independent-samples t-test showed a statistically significant difference in height between the groups, 
# t = 28.05, p < 0.001. Levene's test indicated no significant difference in variances (p = 0.543), and 
# Cohen's d = 1.98 indicated a very large effect size.
```

```python
import numpy as np
from scipy.stats import ttest_ind, levene


def compare_by_sex(df, variable):
    
    # Separate male and female groups
    male = df[df["Sex"] == "Male"][variable].dropna()
    female = df[df["Sex"] == "Female"][variable].dropna()
    
    # Descriptive statistics
    male_mean = male.mean()
    female_mean = female.mean()
    male_sd = male.std()
    female_sd = female.std()
    
    # Levene's test
    levene_stat, levene_p = levene(male, female)
    
    # Choose appropriate t-test
    if levene_p > 0.05:
        t_stat, p_value = ttest_ind(male, female, equal_var=True)
        test_used = "Independent t-test"
    else:
        t_stat, p_value = ttest_ind(male, female, equal_var=False)
        test_used = "Welch's t-test"
    
    # Cohen's d
    pooled_sd = np.sqrt(
        ((len(male) - 1) * male.var(ddof=1) +
         (len(female) - 1) * female.var(ddof=1))
        / (len(male) + len(female) - 2)
    )
    
    cohens_d = (male_mean - female_mean) / pooled_sd
    
    return {
        "Variable": variable,
        "Female Mean": float(female_mean),
        "Female SD": female_sd,
        "Male Mean": male_mean,
        "Male SD": male_sd,
        "Levene p": levene_p,
        "Test": test_used,
        "t-statistic": t_stat,
        "p-value": p_value,
        "Cohen's d": cohens_d
    }
```

```python
result = compare_by_sex(df, "Foot_Breadth_cm")

result
```

```python
variables = [
    "Height_cm",
    "Weight_kg",
    "BMI",
    "Foot_Length_cm",
    "Foot_Breadth_cm",
    "Instep_Length_cm",
    "Heel_Breadth_cm",
    "Bimalleolar_Breadth_cm",
    "Ankle_Height_cm",
    "Calf_Height_cm",
    "Medial_Lateral_Malleolus_Height_cm",
    "Calf_Circumference_cm",
    "Ankle_Circumference_cm",
    "Heel_Ankle_Circumference_cm",
    "Instep_Circumference_cm",
    "Ball_of_Foot_Circumference_cm"
]

results = []

for variable in variables:
    results.append(compare_by_sex(df, variable))
```

```python
results_df = pd.DataFrame(results)

results_df
```

```python
X = df[
    [
        "Height_cm",
        "Weight_kg",
        "BMI",
        "Age_years"
    ]
]

y = df["Foot_Length_cm"]

model = LinearRegression()
model.fit(X, y)
```

```python
y_pred = model.predict(X)

print(y_pred[:10])
```

```python
# On average, how far are our predictions from the actual foot lengths?
# MAE means Mean Absolute Error.

from sklearn.metrics import mean_absolute_error

mae = mean_absolute_error(y, y_pred)

print("MAE:", mae)
```

```python
# Calculate RMSE
from sklearn.metrics import mean_squared_error
import numpy as np

rmse = np.sqrt(mean_squared_error(y, y_pred))

print("RMSE:", rmse)
```
